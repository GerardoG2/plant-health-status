```python
# Functions to Read Data

SENSOR_DATA_PATH = '../Data/Trial_2_Sensor_Data.xlsx'

def load_env_sensor_day_data():
    return pd.read_excel(SENSOR_DATA_PATH, header=2, usecols="K:R", nrows=236)

def load_env_sensor_night_data():
    return pd.read_excel(SENSOR_DATA_PATH, header=2, usecols="A:H", nrows=119)

def load_nutr_sensor_data():
    return pd.read_excel(SENSOR_DATA_PATH, sheet_name='Nutrient Data', header=1, usecols = "A:I", nrows=355)
```
